```python
# Problem 1 — Valid Anagram
# Pattern: Frequency map / hash map

# Logic - anagram problem
#
#1 Different lengths returns false 
#2 Count every character in s 
#3a Go through t and subtract each character 
#3b If character in s is not in t > return false 
#4 If count goes below 0 > t has more characters > return false 
#5 Otherwise return true 

def is_anagram(s, t): 
    if len(s) != len(t): 
        return False
    
    freq = {}

    for char in s: 
        freq[char] = freq.get(char, 0) + 1

    for char in t: 
        if char not in s:
            return False

        freq[char] -= 1

        if freq[char] < 0:
            return False

    return True

print(is_anagram("anagram", "nagaram"))  # True
print(is_anagram("rat", "car"))          # False
```

```python
# Problem 2 — Move Zeroes
# Pattern: Two pointers
# Given an integer array nums, move all 0s to the end while maintaining the relative order of the non-zero elements.

#Logic  (with current understanding)
# 
#1 loop through list and get total count 
#2 loop through list and locate position of zeros 
#3 Move the located numbers to last - with built in function

#Working Logic 
# 
#1 Initialise insert_pos as 0 
#2a Loop through nums from left to right 
#2b If the current number is not 0, move it to nums[insert_pos]
#2c Move insert_pos forward by 1 
#3 Loop till list length and add zeros for remaining positions 


nums = [0, 1, 0, 3, 12]
print(nums)

def move_zeroes(nums):
    insert_pos = 0

    for i in range(len(nums)):
        if nums[i] != 0:
            nums[insert_pos] = nums[i]
            insert_pos += 1

    while insert_pos < len(nums):
        nums[insert_pos] = 0
        insert_pos += 1

move_zeroes(nums)
print(nums)
```

```python
# Problem 3 — Maximum Sum Subarray of Size K
# Pattern: Sliding window
# Given an array of integers nums and an integer k, return the maximum sum of any contiguous subarray of length k.

#Working logic 
# 
#1a Find the sum of the first k numbers 
#1b Store that as both current_sum and max_sum 
#2a Move the window one position to the right 
#2b Subtract the value leaving the window 
#2c Add the value entering the window 
#3 Update the max_sum if new window sum is larger 
#4 Return max_sum 

def max_sum_k(nums, k):
    current_sum = sum(nums[:k])
    max_sum = current_sum

    for i in range(k, len(nums)):
        current_sum -= nums[i-k]
        current_sum += nums[i]

        max_sum = max(max_sum, current_sum)

    return max_sum

print(max_sum_k([2, 1, 5, 1, 3, 2], 3))  # 9
print(max_sum_k([2, 3, 4, 1, 5], 2))     # 7
```

```python
# Problem 4 — Valid Parentheses
# Pattern: Stack

#Logic 
#
#1 Create an empty stack 
#2 Create a mapping for each closing bracket with the appropriate opening bracket 
#3a Go through each character in the string 
#3b If open bracket, add it to the stack 
#4a Else return false, if stack is empty 
#4b Or last element doesnt match with the pairs map, return false 
#5 If it matches, remove opening bracket from the stack 
#6 Loop through the string, return True when the stack is empty 

def valid_parentheses(s):
    stack = []

    pairs_map = {
        ')':'(',
        ']':'[',
        '}':'{'
    }

    for char in s: 
        if char in '([{':
            stack.append(char)

        else: 
            if not stack:
                return False

            if stack[-1] != pairs_map[char]:
                return False

            stack.pop()

    return len(stack) == 0


print(valid_parentheses("()[]{}"))  # True
print(valid_parentheses("([{}])"))  # True
print(valid_parentheses("(]"))      # False
print(valid_parentheses("([)]"))    # False
```

```python
# Problem 5 — Binary Search
# Pattern: Binary search
# Given a sorted list nums and a target value, return the index of the target.

#Logic
# 
#1 Start with two pointers, left at the beginning and right at the end 
#2 While left is less than or equal to right, find the middle index
#3a If nums[mid] is the target, return mid 
#3b If nums[mid] is less than target, move left to mid + 1 
#3c If nums[mid] is high than target, move right to mid - 1 
#4 If the loop ends without finding the target, return -1 

def binary_search(nums, target):
    left = 0
    right = len(nums) - 1 

    while left <= right: 
        mid = (left + right) // 2

        if nums[mid] == target:
            return mid

        if nums[mid] < target: 
            left = mid + 1

        else: 
            right = mid - 1

    return -1


print(binary_search([-1, 0, 3, 5, 9, 12], 9))  # 4
print(binary_search([-1, 0, 3, 5, 9, 12], 2))  # -1
```
