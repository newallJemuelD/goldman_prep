```python
# Problem 1 — Reverse Linked List
# Pattern: Linked list / pointer manipulation
# Given the head of a singly linked list, reverse the list and return the new head.

#Logic 
# 1. Start previous as None
#    because the new tail should point to nothing
# 2. Start current at head
# 3. While current is not None:
#    save current.next before changing it
# 4. Reverse the pointer:
#    current.next should point to previous
# 5. Move previous forward to current
# 6. Move current forward to the saved next node
# 7. When the loop ends,
#    previous is the new head
# 8. Return previous

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next


def reverse_list(head):
    previous = None
    current = head

    while current:
        next_node = current.next

        current.next = previous

        previous = current
        current = next_node

    return previous


head = ListNode(1)
head.next = ListNode(2)
head.next.next = ListNode(3)
head.next.next.next = ListNode(4)
head.next.next.next.next = ListNode(5)

new_head = reverse_list(head)

current = new_head

while current:
    print(current.val, end=" -> ")
    current = current.next

print("None")
```

```python
# Problem 2 — Maximum Depth of Binary Tree
# Pattern: Tree DFS / recursion
# Given the root of a binary tree, return its maximum depth.

#Logic 
# 1. If the tree is empty, return 0
# 2. Recursively find the maximum depth of the left subtree
# 3. Recursively find the maximum depth of the right subtree
# 4. Take the larger of the two depths
# 5. Add 1 for the current node
# 6. Return the result

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def max_depth(root):
    if root is None:
        return 0

    left_depth = max_depth(root.left)
    right_depth = max_depth(root.right)

    return 1 + max(left_depth, right_depth)


root = TreeNode(3)

root.left = TreeNode(9)
root.right = TreeNode(20)

root.right.left = TreeNode(15)
root.right.right = TreeNode(7)

print(max_depth(root))  # 3
```

```python
# Problem 3 — Climbing Stairs
# Pattern: Dynamic programming
# You are climbing a staircase with n steps.

#Logic 
# 1. If n is 1, there is only 1 way
# 2. If n is 2, there are 2 ways
# 3. Track the number of ways to reach the previous two steps
# 4. For each new step:
#    current = previous_one + previous_two
# 5. Move the two stored values forward
# 6. Continue until step n
# 7. Return the number of ways for step n

def climb_stairs(n):
    if n == 1:
        return 1

    if n == 2:
        return 2

    previous_two = 1
    previous_one = 2

    for i in range(3, n + 1):
        current = previous_one + previous_two

        previous_two = previous_one
        previous_one = current

    return previous_one

print(climb_stairs(2))  # 2
print(climb_stairs(3))  # 3
print(climb_stairs(4))  # 5
print(climb_stairs(5))  # 8
```

```python
# Problem 4 — Shortest Path in a Grid
# Pattern: BFS ( Breadth First Search )

#Logic 
# 1. If the grid is empty, return -1
# 2. If the start or destination cell is blocked,
#    return -1
# 3. Create a queue for BFS
# 4. Put the starting cell (0, 0) in the queue
#    together with distance 0
# 5. Mark the starting cell as visited
# 6. Remove cells from the queue one by one
# 7. If the current cell is the bottom-right cell,
#    return its distance
# 8. Check its 4 neighbors:
#    up, down, left, right
# 9. If a neighbor is inside the grid,
#    is open, and has not been visited:
#    - mark it visited
#    - add it to the queue with distance + 1
# 10. If BFS finishes without reaching the destination,
#     return -1


from collections import deque

def shortest_path(grid):
    if not grid:
        return -1

    rows = len(grid)
    cols = len(grid[0])

    if grid[0][0] == 1 or grid[rows - 1][cols - 1] == 1:
        return -1

    queue = deque([(0, 0, 0)])
    visited = {(0, 0)}

    directions = [
        (-1, 0),  # up
        (1, 0),   # down
        (0, -1),  # left
        (0, 1)    # right
    ]

    while queue:
        r, c, distance = queue.popleft()

        if r == rows - 1 and c == cols - 1:
            return distance

        for dr, dc in directions:
            new_r = r + dr
            new_c = c + dc

            if (
                0 <= new_r < rows and
                0 <= new_c < cols and
                grid[new_r][new_c] == 0 and
                (new_r, new_c) not in visited
            ):
                visited.add((new_r, new_c))
                queue.append((new_r, new_c, distance + 1))

    return -1


grid = [
    [0, 0, 1],
    [1, 0, 0],
    [1, 1, 0]
]
print(shortest_path(grid))  # 4
```

```python
# Problem 5 — Non-overlapping Meetings
# Pattern: Greedy / interval scheduling

#Logic
# 1. If there are no meetings, return 0
# 2. Sort meetings by their end time
# 3. Select the meeting that finishes earliest
# 4. Track the end time of the last selected meeting
# 5. Go through the remaining meetings
# 6. If a meeting starts at or after the previous
#    selected meeting ends, select it
# 7. Increase the meeting count and update
#    the last end time
# 8. Return the total number of selected meetings

def max_meetings(meetings):
    if not meetings:
        return 0

    meetings.sort(key=lambda x: x[1])

    count = 1
    last_end = meetings[0][1]

    for start, end in meetings[1:]:
        if start >= last_end:
            count += 1
            last_end = end

    return count

meetings = [
    [1, 3],
    [2, 4],
    [3, 5],
    [6, 8]
]
print(max_meetings(meetings))  # 3
```
