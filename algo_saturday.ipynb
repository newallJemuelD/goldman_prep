```python
# Problem 1 — Range Sum Query
# Pattern: Prefix Sum
# You are given an integer array nums.
# You will receive multiple queries. Each query asks for the sum of the elements between indices left and right, inclusive.

#Logic 
# 
#1 Create a prefix array starting with 0
#2 Go through nums from left to right
#3 For each number, add it to the previous prefix total
#4 prefix[i] should represent the sum of elements, before index i
#5 To get the sum from left to right inclusive: subtract prefix[left] from prefix[right + 1]
#6 Return the result


def build_prefix_sum(nums):
    prefix = [0]

    for num in nums:
        prefix.append(prefix[-1] + num)

    return prefix


def range_sum(prefix, left, right):
    return prefix[right + 1] - prefix[left]


nums = [2, 4, 1, 7, 3]
prefix = build_prefix_sum(nums)

print(prefix)                  # [0, 2, 6, 7, 14, 17]
print(range_sum(prefix, 1, 3)) # 12
print(range_sum(prefix, 0, 4)) # 17
```

```python
# Problem 2 — Kth Largest Element
# Pattern: Heap / Priority Queue
# Given an integer array nums and an integer k, return the kth largest element.

#Logic
# 1. Create an empty min-heap
# 2. Go through each number in nums
# 3. Push the current number into the heap
# 4. If the heap size becomes greater than k, remove the smallest element
# 5. Keep doing this so the heap always contains only the k largest values seen so far
# 6. At the end, the smallest value in the heap is the kth largest value overall
# 7. Return heap[0]

import heapq

def find_kth_largest(nums, k):
    heap = []

    for num in nums:
        heapq.heappush(heap, num)

        if len(heap) > k:
            heapq.heappop(heap)

    return heap[0]

print(find_kth_largest([3, 2, 1, 5, 6, 4], 2))  # 5
print(find_kth_largest([3, 2, 3, 1, 2, 4, 5, 5, 6], 4))  # 4
```

```python
# Problem 3 — Subsets
# Pattern: Recursion / Backtracking
# Given an array of unique integers nums, return all possible subsets.

#Logic 
# 1. Start with an empty result list
# 2. Use a helper function with:
#    - the current index
#    - the current subset
# 3. At each index, make two choices:
#    - do not include nums[index]
#    - include nums[index]
# 4. Move to the next index after each choice
# 5. When index reaches the end of nums,
#    add a copy of the current subset to the result
# 6. When backtracking, remove the last added element
#    before exploring the next choice
# 7. Return all collected subsets


def subsets(nums):
    result = []

    def backtrack(index, current):
        if index == len(nums):
            result.append(current.copy())
            return

        # Do not include nums[index]
        backtrack(index + 1, current)

        # Include nums[index]
        current.append(nums[index])
        backtrack(index + 1, current)

        # Backtrack
        current.pop()

    backtrack(0, [])

    return result



print(subsets([1, 2]))
# [[], [2], [1], [1, 2]]
print(subsets([1, 2, 3]))
# 8 subsets
```

```python
# Problem 4 — Number of Islands
# Pattern: BFS / DFS
# You are given a 2D grid containing:

#Logic 
# 1. Start island_count at 0
# 2. Go through every cell in the grid
# 3. When you find a "1":
#    increase island_count by 1
# 4. Run DFS from that cell
# 5. DFS should stop if:
#    - row is outside the grid
#    - column is outside the grid
#    - the cell is not land
# 6. Mark the current land cell as visited
#    by changing "1" to "0"
# 7. Continue DFS in 4 directions:
#    up, down, left, right
# 8. After the whole grid is checked,
#    return island_count


def num_islands(grid):
    if not grid:
        return 0

    rows = len(grid)
    cols = len(grid[0])
    island_count = 0

    def dfs(r, c):
        if (
            r < 0 or
            r >= rows or
            c < 0 or
            c >= cols or
            grid[r][c] != "1"
        ):
            return

        grid[r][c] = "0"

        dfs(r - 1, c)  # up
        dfs(r + 1, c)  # down
        dfs(r, c - 1)  # left
        dfs(r, c + 1)  # right

    for r in range(rows):
        for c in range(cols):
            if grid[r][c] == "1":
                island_count += 1
                dfs(r, c)

    return island_count

grid = [
    ["1", "1", "0", "0"],
    ["1", "0", "0", "1"],
    ["0", "0", "1", "1"]
]

print(num_islands(grid))  # 2
```

```python
# Problem 5 — Merge Intervals
# Pattern: Sorting + Intervals
# You are given a list of intervals:

#Logic 
# 1. If intervals is empty, return []
# 2. Sort the intervals by their start value
# 3. Put the first interval into the result list
# 4. Go through the remaining intervals one by one
# 5. Compare the current interval with the last interval in result
# 6. If current_start <= previous_end,
#    the intervals overlap, so merge them
# 7. Update the previous end to the larger of:
#    previous_end or current_end
# 8. If they do not overlap,
#    add the current interval to result
# 9. Return the merged result


def merge_intervals(intervals):
    if not intervals:
        return []

    intervals.sort(key=lambda x: x[0])

    result = [intervals[0]]

    for current in intervals[1:]:
        current_start = current[0]
        current_end = current[1]

        previous_end = result[-1][1]

        if current_start <= previous_end:
            result[-1][1] = max(previous_end, current_end)
        else:
            result.append(current)

    return result


print(merge_intervals([
    [1, 3],
    [2, 6],
    [8, 10],
    [15, 18]
]))
# [[1, 6], [8, 10], [15, 18]]
print(merge_intervals([[1, 4], [4, 5]]))
# [[1, 5]]
```
