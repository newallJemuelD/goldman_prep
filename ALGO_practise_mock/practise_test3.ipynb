```python
# Problem 1 — Count Numbers Above Average
# Builds on: loops, counters, list traversal
# Given a list of integers, return how many values are greater than the average of the list.

# Main concepts: sum(), len(), average, for loop, comparison, counter

def count_above_average(nums):
    if not nums:
        return 0

    values_counter = 0
    average = sum(nums) / len(nums)

    for num in nums: 
        if num > average:
            values_counter += 1
    
    return values_counter

print(count_above_average([0,1,2,3]))
```

```python
# Problem 2 — First Unique Character
# Builds on: character frequency / hash map
# Given a string, return the first character that appears exactly once.
# If every character repeats, return None.

# Main concepts: dictionary / hash map, .get(), for loop, frequency counting, 
# dictionary lookup, second pass, early return

def first_unique(text):
    char_freq = {}

    for char in text: 
        char_freq[char] = char_freq.get(char, 0) + 1

    for char in text: 
        if char_freq[char] == 1:
            return char

    return None


print(first_unique("banana"))
```

```python
# Problem 3 — Pair Difference
# Builds on: set + complement reasoning
# Given a list of integers and a number k, return True if two different values have an absolute difference of k.

# Main concepts: set(), membership, absolute difference, early return

def has_difference(nums, k):
    seen = set()

    for num in nums:
        if num - k in seen or num + k in seen:
            return True

        seen.add(num)

    return False

print(has_difference([0,1,2,3], 1))
```

```python
# Problem 4 — Longest Consecutive Streak of the Same Value
# Builds on: counters + tracking maximum
# Given a list of integers, return the length of the longest run of identical consecutive values.

# Concepts: indexing, range(), comparison, counter reset, max tracking

def longest_streak(nums):
    if not nums:
        return 0

    current_streak = 1
    max_streak = 1

    for i in range(1, len(nums)):
        if nums[i] == nums[i-1]:
            current_streak += 1
            max_streak = max(max_streak, current_streak)

        else: 
            current_streak = 1

    return max_streak


print(longest_streak([1, 1, 2, 2, 2, 3, 3]))
```

```python
# Problem 5 — Two Sum Indices
# Builds directly on: your previous pair-sum problem
# Now return the indices of the two values whose sum equals target.

def two_sum(nums, target):
    seen = {}

    for i,num in enumerate(nums): 
        needed = target - num

        if needed in seen:
            return [seen[needed], i]
        
        seen[num] = i

print(two_sum([2, 7, 11, 15], 9))
```
