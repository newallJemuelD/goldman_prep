```python
# Problem 1 — FizzBuzz With a Counter
# Given an integer n, examine every number from 1 to n.
# - divisible by both 3 and 5 → "FizzBuzz"
# - divisible by 3 → "Fizz"
# - divisible by 5 → "Buzz"
# - otherwise → the number
# Instead of printing the values, return how many numbers belong to each category.

def fizzbuzz_counter(number):
    counter_dictionary = {
        "Fizz" : 0,
        "Buzz" : 0,
        "FizzBuzz" : 0,
        "Number" : 0
    }

    for num in range(1, number+1):
        if num % 3 == 0 and num % 5 == 0:
            counter_dictionary["FizzBuzz"] += 1
        elif num % 3 == 0:
            counter_dictionary["Fizz"] += 1
        elif num % 5 == 0:
            counter_dictionary["Buzz"] += 1
        else: 
            counter_dictionary["Number"] += 1

    return counter_dictionary


print(fizzbuzz_counter(15))
```

```python
# Problem 2 — First Repeated Number
# Given a list of integers, return the first number that appears for a second time.
# If there is no repeated number, return -1.

# Main concepts: set(), membership lookup, for loop, early return

def first_repeat(nums):
    unique_nums = set()

    for num in nums: 
        if num in unique_nums: 
            return num 

        unique_nums.add(num)

    return -1

print(first_repeat([0,1,2,3,3]))
```

```python
# Problem 3 — Most Frequent Character
# Given a string, return the character that occurs most frequently.
# Assume there is a unique most frequent character.

# Main concepts: dictionary / hash map, frequency counting, loop through dictionary, track best value

def most_frequent(word):
    freq_dictionary = {}
    best_char = None
    best_count = 0

    for char in word:
        if char in freq_dictionary: 
            freq_dictionary[char] += 1

        else:   
            freq_dictionary[char] = 1

    for char, count in freq_dictionary.items():
        if count > best_count:
            best_count = count
            best_char = char

    return best_char
        

print(most_frequent("banana"))
```

```python
# Problem 4 — Remove Duplicates While Keeping Order
# Given a list of integers, return a new list containing each value only once.
# Keep the values in the order in which they first appeared.

# Main concepts: set, list, membership check, append()

def remove_duplicates(nums):
    seen = set()
    result = []

    for num in nums:
        if num not in seen:
            seen.add(num)
            result.append(num)

    return result

print(remove_duplicates([0,1,3,1,1]))
```

```python
# Problem 5 — Pair With Target Sum
# Given a list of integers and a target, return True if two different elements add up to the target.
# Otherwise return False.

# Main concepts: set, complement, target - current number, membership lookup

def has_pair(nums, target):
    seen = []

    for num in nums: 
        needed = target - num

        if needed in seen:
            return True

        else: 
            seen.append(num)

    return False

has_pair([0,1,2], 3)
```
