```python
# Question 1 — Number Labels
# Given an integer n, generate a list containing the values from 1 to n.
# For each number:
# - if it is divisible by both 3 and 5, add "AlphaBeta"
# - if it is divisible only by 3, add "Alpha"
# - if it is divisible only by 5, add "Beta"
# - otherwise add the number itself as a string
# Return the resulting list.

def number_labels(n):
    result = []

    for num in range(1, n+1):
        if num % 3 == 0 and num % 5 == 0:
            result.append("AlphaBeta")
        elif num % 3 == 0: 
            result.append("Alpha")
        elif num % 5 == 0:
             result.append("Beta")
        else:
            result.append(str(num))

    return result

print(number_labels(7))
```

```python
# Question 2 — First Duplicate Within Distance K
# You are given a list of integers nums and an integer k.
# Return True if the same number appears twice such that the two occurrences are at most k positions apart.
# Otherwise return False.

def nearby_duplicate(nums, k):
    last_seen = {}

    for i, num in enumerate(nums):
        if num in last_seen:
            distance = i - last_seen[num]

            if distance <= k:
                return True

        last_seen[num] = i

    return False

print(nearby_duplicate([1, 2, 3, 1], 3))     # True
print(nearby_duplicate([1, 2, 3, 1], 2))     # False
print(nearby_duplicate([5, 7, 5, 8, 7], 2))  # True
print(nearby_duplicate([], 3))                # False
```
