```python
# Question 1 — First Unique Number
# You are given a list of integers nums.
# Return the first number that occurs exactly once in the list.
# If every number appears more than once, return -1.

def first_unique(nums):
    frequency = {}

    for num in nums:
        frequency[num] = frequency.get(num, 0) + 1

    for num in nums:
        if frequency[num] == 1:
            return num

    return -1

print(first_unique([1,1,2,1,2,8]))
```

```python
# Question 2 — Maximum Sum of K Consecutive Elements

def max_window_sum(nums, k):
    current_sum = sum(nums[:k])
    max_sum = current_sum

    for i in range(k, len(nums)):
        current_sum -= nums[i - k]
        current_sum += nums[i]

        max_sum = max(max_sum, current_sum)

    return max_sum

print(max_window_sum([2, 1, 5, 1, 3, 2], 3))  # 9
```
