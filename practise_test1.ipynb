```python
# Problem 1 — Multiples of 4
# Given an integer n, print every number from 1 to n.
# - If the number is divisible by 4, print "Four"
# - Otherwise print the number

def multiples_of_four(n):
    for number in range(1, n+1):
        if number % 4 == 0: 
            print('Four')

        else: 
            print(number)


multiples_of_four(7)
```

```python
# Problem 2 — Count Positive and Negative Numbers
# Given a list of integers, return the number of positive and negative values.
# Ignore zeros.

def count_signs(numbers_list):
    positive_count = 0
    negative_count = 0

    for number in numbers_list:
        if number > 0:
            positive_count += 1

        elif number < 0:
            negative_count += 1

    return print(positive_count, negative_count)

count_signs([-2, -1, 0, 1, 2])
```

```python
# Problem 3 — Reverse a String
# Given a string, return the string reversed.

# Method 1: Using list, stack to reverse 
def reverse_string(word):
    chars = []

    for char in word: 
        chars.append(char)

    result = ""

    while chars:
        result += chars.pop()

    return result

print(reverse_string("newall"))


## Method 2: Reverse string variable
# def reverse_string(text):
#     reversed_text = ""

#     for char in text:
#         reversed_text = char + reversed_text

#     return reversed_text

# print(reverse_string("newall"))


# # Method 3: Slicing function 
# def reverse_string(text):
#     return text[::-1]

# print(reverse_string("newall"))
```

```python
# Problem 4 — Count Character Frequency
# Given a string, return a dictionary showing how many times each character occurs.

# Method 1: Using a loop and dictionary 
def char_counter(word):
    char_dictionary = {}
    
    for char in word: 
        if char in char_dictionary: 
            char_dictionary[char] += 1
        else: 
            char_dictionary[char] = 1

    return char_dictionary

print(char_counter("banana"))

# # Method 2: Using inbuilt get method of dictionary 
# def character_count(text):
#     counts = {}

#     for char in text:
#         counts[char] = counts.get(char, 0) + 1

#     return counts

# print(char_counter("banana"))
```
