```python
# Problem 1 — Group Anagrams
# Given an array of strings, group the anagrams together.

#Logic 
# 1. Create an empty dictionary for groups
# 2. Go through each word in strs
# 3. Sort the characters in the word
# 4. Join the sorted characters into a string
#    and use it as the dictionary key
# 5. If the key is not already in the dictionary,
#    create an empty list for it
# 6. Add the original word to that key's group
# 7. Return all dictionary values as a list

def group_anagrams(strs):
    groups = {}

    for word in strs:
        key = "".join(sorted(word))

        if key not in groups:
            groups[key] = []

        groups[key].append(word)

    return list(groups.values())

print(group_anagrams(
    ["eat", "tea", "tan", "ate", "nat", "bat"]
))
```

```python
# Problem 2 — Binary Tree Level Order Traversal
# Given the root of a binary tree, return the values level by level.

#Logic
# 1. If root is None, return an empty list
# 2. Create a queue and put the root node in it
# 3. Create an empty result list
# 4. While the queue is not empty:
#    record how many nodes are currently in the queue
#    this is the number of nodes in the current level
# 5. Process exactly level_size nodes
# 6. Add each node's value to the current level list
# 7. Add the node's left child to the queue if it exists
# 8. Add the node's right child to the queue if it exists
# 9. After finishing that level,
#    add the level list to the result
# 10. Return the result

from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def level_order(root):
    if root is None:
        return []

    result = []
    queue = deque([root])

    while queue:
        level_size = len(queue)
        level = []

        for _ in range(level_size):
            node = queue.popleft()

            level.append(node.val)

            if node.left:
                queue.append(node.left)

            if node.right:
                queue.append(node.right)

        result.append(level)

    return result


root = TreeNode(3)
root.left = TreeNode(9)
root.right = TreeNode(20)
root.right.left = TreeNode(15)
root.right.right = TreeNode(7)
print(level_order(root))
# [[3], [9, 20], [15, 7]]
```

```python
# Problem 3 — House Robber
# You are given an array where each value represents the amount of money in a house.
# You cannot rob two adjacent houses.
# Return the maximum amount of money you can rob.

#Logic
# 1. If nums is empty, return 0
# 2. Track the best result from:
#    - two houses ago
#    - the previous house
# 3. For each house, choose the better of:
#    - skip current house → previous_best
#    - rob current house → current_value + two_steps_back
# 4. Store that as current_best
# 5. Move the two tracked values forward
# 6. After processing all houses,
#    return previous_best

def rob(nums):
    two_steps_back = 0
    previous_best = 0

    for value in nums:
        current_best = max(
            previous_best,
            value + two_steps_back
        )

        two_steps_back = previous_best
        previous_best = current_best

    return previous_best

print(rob([1, 2, 3, 1]))      # 4
print(rob([2, 7, 9, 3, 1]))   # 12
print(rob([]))                 # 0
```

```python
# Problem 4 — Network Delay Time
# You are given a directed weighted graph.

#Logic
# 1. Build a graph where each node stores:
#    (neighbor, travel_time)
# 2. Use a min-heap starting with:
#    (0, k)
#    meaning distance 0 from the starting node
# 3. Keep a dictionary of shortest confirmed distances
# 4. Always remove the node with the smallest distance
#    from the heap
# 5. If that node was already processed,
#    skip it
# 6. Otherwise, record its shortest distance
# 7. For each neighbor:
#    new_distance = current_distance + edge_weight
# 8. Push the neighbor and new distance into the heap
# 9. After processing:
#    if not all n nodes were reached → return -1
# 10. Otherwise, return the largest shortest distance
#     because that is when the last node receives the signal

import heapq

def network_delay_time(times, n, k):
    graph = {}

    for u, v, time in times:
        if u not in graph:
            graph[u] = []

        graph[u].append((v, time))

    heap = [(0, k)]
    distances = {}

    while heap:
        current_time, node = heapq.heappop(heap)

        if node in distances:
            continue

        distances[node] = current_time

        for neighbor, travel_time in graph.get(node, []):
            if neighbor not in distances:
                new_time = current_time + travel_time
                heapq.heappush(heap, (new_time, neighbor))

    if len(distances) != n:
        return -1

    return max(distances.values())


times = [
    [2, 1, 1],
    [2, 3, 1],
    [3, 4, 1]
]
print(network_delay_time(times, 4, 2))  # 2
```

```python
# Problem 5 — Meeting Rooms II
# You are given meeting intervals.
# Return the minimum number of meeting rooms required so that no meetings overlap in the same room.

#Logic
# 1. If there are no intervals, return 0
# 2. Sort meetings by start time
# 3. Create a min-heap to store meeting end times
# 4. Add the first meeting's end time to the heap
# 5. Go through the remaining meetings
# 6. If the current meeting starts at or after
#    the earliest-ending meeting finishes:
#    remove that end time from the heap
#    because that room can be reused
# 7. Add the current meeting's end time to the heap
# 8. The heap size represents how many rooms
#    are currently needed
# 9. Return the final heap size

import heapq

def min_meeting_rooms(intervals):
    if not intervals:
        return 0

    intervals.sort(key=lambda x: x[0])

    heap = []

    heapq.heappush(heap, intervals[0][1])

    for start, end in intervals[1:]:
        if start >= heap[0]:
            heapq.heappop(heap)

        heapq.heappush(heap, end)

    return len(heap)


print(min_meeting_rooms([
    [0, 30],
    [5, 10],
    [15, 20]
]))
# 2
print(min_meeting_rooms([
    [7, 10],
    [2, 4]
]))
# 1
```
