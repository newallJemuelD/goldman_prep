```python
# Problem 1 — Subarray Sum Equals K
# Pattern: Prefix sum + hash map
# Given an integer array nums and an integer k, return the number of contiguous subarrays whose sum equals k.

#Logic
# 1. Keep a running sum as you move through nums
# 2. Store how many times each prefix sum has appeared
# 3. Start with prefix sum 0 appearing once
#    so subarrays starting at index 0 can be counted
# 4. For each number, update current_sum
# 5. Calculate current_sum - k
# 6. If current_sum - k has appeared before,
#    add its frequency to the answer
# 7. Store the current_sum in the hash map
# 8. Return the total count

def subarray_sum(nums, k):
    prefix_count = {0: 1}

    current_sum = 0
    count = 0

    for num in nums:
        current_sum += num

        needed = current_sum - k

        if needed in prefix_count:
            count += prefix_count[needed]

        prefix_count[current_sum] = prefix_count.get(current_sum, 0) + 1

    return count

print(subarray_sum([1, 1, 1], 2))  # 2
print(subarray_sum([1, 2, 3], 3))  # 2
```

```python
# Problem 2 — Top K Frequent Elements
# Pattern: Hash map + heap
# Given an integer array nums and integer k, return the k most frequent elements.

#Logic 
# 
# 1. Count how many times each number appears
# 2. Create an empty min-heap
# 3. For each number and its frequency:
#    push (frequency, number) into the heap
# 4. If the heap grows larger than k,
#    remove the item with the smallest frequency
# 5. After processing all numbers,
#    the heap contains the k most frequent elements
# 6. Return only the numbers from the heap

from collections import Counter
import heapq

def top_k_frequent(nums, k):
    freq = Counter(nums)
    heap = []

    for num, count in freq.items():
        heapq.heappush(heap, (count, num))

        if len(heap) > k:
            heapq.heappop(heap)

    return [num for count, num in heap]


print(top_k_frequent([1, 1, 1, 2, 2, 3], 2))
# [2, 1] or [1, 2]
print(top_k_frequent([1], 1))
# [1]
```

```python
# Problem 3 — Clone Graph
# Pattern: DFS / graph traversal
# You are given a node in a connected undirected graph.
# Each node has:
# node.valnode.neighbors
# Return a deep copy of the entire graph.

#Logic
# 1. If node is None, return None
# 2. Create a dictionary called clones
# 3. The dictionary maps:
#    original node -> cloned node
# 4. Use DFS to visit each node
# 5. If a node was already cloned,
#    return the existing clone
# 6. Otherwise, create a new clone
#    and store it in the dictionary immediately
# 7. Visit each neighbor of the original node
# 8. Clone each neighbor recursively
#    and add it to the cloned node's neighbors
# 9. Return the cloned starting node


class Node:
    def __init__(self, val=0, neighbors=None):
        self.val = val
        self.neighbors = neighbors if neighbors is not None else []


def clone_graph(node):
    if node is None:
        return None

    clones = {}

    def dfs(current):
        if current in clones:
            return clones[current]

        clone = Node(current.val)
        clones[current] = clone

        for neighbor in current.neighbors:
            clone.neighbors.append(dfs(neighbor))

        return clone

    return dfs(node)

# Build this graph:
#
# 1 -- 2
# |    |
# 4 -- 3

node1 = Node(1)
node2 = Node(2)
node3 = Node(3)
node4 = Node(4)

node1.neighbors = [node2, node4]
node2.neighbors = [node1, node3]
node3.neighbors = [node2, node4]
node4.neighbors = [node1, node3]

cloned = clone_graph(node1)

print(cloned.val)  # 1
print([neighbor.val for neighbor in cloned.neighbors])  # [2, 4]
```

```python
# Problem 4 — Insert Interval
# Pattern: Intervals
# You are given a sorted list of non-overlapping intervals and one new interval.
# Insert the new interval and merge if necessary.

#Logic
# 1. Create an empty result list
# 2. Add all intervals that end before the new interval starts
# 3. While intervals overlap with the new interval:
#    merge them by updating:
#    - new start = minimum start
#    - new end = maximum end
# 4. Add the merged new interval to the result
# 5. Add all remaining intervals that come after it
# 6. Return the result

def insert_interval(intervals, new_interval):
    result = []
    i = 0

    # Add intervals completely before new_interval
    while i < len(intervals) and intervals[i][1] < new_interval[0]:
        result.append(intervals[i])
        i += 1

    # Merge overlapping intervals
    while i < len(intervals) and intervals[i][0] <= new_interval[1]:
        new_interval[0] = min(new_interval[0], intervals[i][0])
        new_interval[1] = max(new_interval[1], intervals[i][1])
        i += 1

    result.append(new_interval)

    # Add intervals completely after new_interval
    while i < len(intervals):
        result.append(intervals[i])
        i += 1

    return result


print(insert_interval(
    [[1, 3], [6, 9]],
    [2, 5]
))
# [[1, 5], [6, 9]]
print(insert_interval(
    [[1, 2], [3, 5], [6, 7], [8, 10], [12, 16]],
    [4, 8]
))
# [[1, 2], [3, 10], [12, 16]]
```

```python
# Problem 5 — Jump Game
# Pattern: Greedy
# Given an array nums, each value tells you the maximum number of positions you can jump forward from that index.
# Return True if you can reach the final index.

#Logic
# 1. Start with furthest = 0
# 2. Go through nums from left to right
# 3. If the current index is greater than furthest,
#    you cannot reach this position, so return False
# 4. Otherwise, update furthest using:
#    max(furthest, i + nums[i])
# 5. If furthest reaches or passes the final index,
#    return True
# 6. If the loop finishes without getting stuck,
#    return True

def can_jump(nums):
    furthest = 0

    for i in range(len(nums)):
        if i > furthest:
            return False

        furthest = max(furthest, i + nums[i])

        if furthest >= len(nums) - 1:
            return True

    return True

print(can_jump([2, 3, 1, 1, 4]))  # True
print(can_jump([3, 2, 1, 0, 4]))  # False
print(can_jump([0]))               # True
```
