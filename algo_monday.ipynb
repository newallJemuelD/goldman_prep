```python
#Question 1 - Contains duplicate
#Pattern: Set
#Given an integer array nums, return True if any value appears at least twice, otherwise return False

def contains_duplicate(nums):
    seen = set()
    for num in nums: 
        if num in seen:
            return True
        seen.add(num)

    return False 

print(contains_duplicate([1, 2, 3, 1]))  # True
print(contains_duplicate([1, 2, 3, 4]))  # False

#---------------------------------------------------------------------------------------------------------
```

```python
# Question 2 - Two sum
# Pattern: Hash map
# Given an array nums and integer target, return the indices of two different elements whose values add to target.

def two_sum(nums, target):
    seen = {}

    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i

print(two_sum([2, 7, 11, 15], 9))

```
